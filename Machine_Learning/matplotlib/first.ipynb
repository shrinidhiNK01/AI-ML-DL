```python
# Matplotlib provides object oriented api for embedding points 
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
# Ensure x and y are the same length
x_plot=np.arange(1,10)
y_plot=np.arange(11,20)

# both should have same size
# scatter the values in 2d graph
plt.scatter(x_plot, y_plot, c="r")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("God is comming")
plt.savefig("snk.png")
```

```python
print(plt.plot(x_plot,y_plot,c="y"))
```

```python
print(plt.plot(x_plot,y_plot,"r--"))
```

```python
print(plt.plot(x_plot,y_plot,"r-*"))
```

```python
print(plt.plot(x_plot,y_plot,"ro"))
```

```python
print(plt.plot(x_plot,y_plot,"ro--"))
```

```python
print(plt.plot(x_plot,y_plot,"r",linestyle="dashed",linewidth=20,markersize=12))
```

```python
# Here 1,2,3 are indexes , at which u wanna keep those subplots
# and u can do all style, like u did above
# 2=rows, 2=coloumns, 1=index
print(plt.subplot(2,2,1))
plt.plot(x_plot,y_plot,c="g")
print(plt.subplot(2,2,2))
plt.plot(x_plot,y_plot,c="r")
print(plt.subplot(2,2,3))
plt.plot(x_plot,y_plot,c="b")
```

```python
x=np.arange(1,20)
y=np.sin(x)
plt.plot(x,y)
```

```python
x=[7,4,3,5,6]
y=[6,7,8,2,4]
x2=[5,2,3,4,5]
y2=[1,7,8,3,1]
plt.bar(x,y,align='center')
plt.bar(x2, y2, color="g")
print(plt.plot(x,y))
plt.title("Bar chart")
plt.xlabel("X-label")
plt.ylabel("y-label")

```

```python
# Histogram
# It specifies density of numbers
a=np.array([11,21,13,41,15,61,71,81,19])
print(plt.hist(a))
plt.title("Histogram")
```

```python
# Boxplot represents percentiles

# Normal std deviation
data=[np.random.normal(0,std,100) for std in range(1,4)]
# patch_artist=True is colur to the graph and vert is verticle, if u remove that, it gives graph in horizontal
plt.boxplot(data,vert=True,patch_artist=True)


print(data)
```

```python
# Pie chart
labels=["python","c++","go","c++","c","java","Javascript"]
sizes=[100,200,300,400,500,600,700]
colors=["yellow","green","blue","red","pink","orange","brown"]
explode=(0.5, 0, 0, 0, 0, 0, 0)

plt.pie(sizes, labels=labels, explode=explode, colors=colors, shadow=True,autopct="%1.1f%%")
plt.axis("equal")
```
