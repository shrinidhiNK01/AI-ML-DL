```python
# helps to analyse various features relationships in our data
# only 1 function=univariate analysis
# 2 function=bivariate analysis
# Distribution plots:-a)displot b)joinplot c)pairplot

import seaborn as sns
df = sns.load_dataset("iris")
print(df)

```

```python
# Heatmap is a coloured monochromatic that show 2d matrix table b/w 2 descrete dimensions   ,important in feature selection
print(df.head())
print(df.corr)
```

```python
# it gives correlation colour, high to low
sns.heatmap(df.select_dtypes(include='number').corr())
```

```python
# Join plot:-study relationship b/w 2 numeric variables ,it is usually a scattern plot, hexbin plot,2d histogram,2d density plot
# univariate analysis
# comparing reationship b/w those 2
sns.jointplot(x="sepal_width",y="petal_length",data=df,kind="hex" )
# dark region is highly concentrated area depects the relation
```

```python
sns.jointplot(x="sepal_width",y="petal_length",data=df,kind="reg" )
# helps to draw regression line, best fit line
# these curve lines are probability density function
```

```python
# Pair plot is a scatter plot, 1 variable in same data row is matched with another variables value
# if we have more than 2 value , it just do permutaion and combination, it combines everything
sns.pairplot(df)
```

```python
sns.pairplot(df, hue="species")
# The following line caused a KeyError because 'sepal_length ' has a trailing space and is not a categorical column.
# If you want to color by a column, use a categorical column like 'species'.
# sns.pairplot(df, hue="sepal_length")  # Not recommended, as 'sepal_length' is numeric.
```

```python
# dist plot:-check distribution of colomns feature
sns.distplot(df["petal_length"])
```

```python
sns.distplot(df["petal_length"],kde=False,bins=10)
```

```python
# Catagorical plots
# '''1)box plot
# 2)violin plot
# 3)Count plot
# 4)Bar plot'''
# The iris dataset does not have a 'sex' column. Let's use 'species' instead.
sns.countplot(x="species", data=df)
# show the counts of observation    
```

```python
sns.barplot(x="species", y="sepal_length", data=df)
```

```python
# boxplot presents information from a five number summary
sns.boxplot(x="petal_length", y="sepal_length", data=df)
```

```python
sns.boxplot(x="petal_length", y="sepal_length", data=df,palette="rainbow")
```

```python
sns.boxplot(data=df, orient="v")
```

```python
# violin plot helps to see both distribution of data interms of kernal density estimation and box plot
sns.violinplot(x="petal_length", y="sepal_length", data=df, palette="rainbow")
```
