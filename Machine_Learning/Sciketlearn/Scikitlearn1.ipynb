```python
from sklearn.preprocessing import StandardScaler  
import numpy as np  
  
data = np.array([[1, 2, 3], [4, 5, 6]], dtype=object)  # Object dtype (mixed types)  
scaler = StandardScaler()  
scaled_data = scaler.fit_transform(data.astype(float))  # Convert to float  
  
print(scaled_data)  
```
